```python
import aar
import numpy as np
from matplotlib import pyplot as plt
from matplotlib.patches import Arc
from importlib import reload
import aar.draw_angle_between_vectors
from matplotlib import quiver
import aar.compute_angles
reload(aar.compute_angles)
reload(aar.draw_angle_between_vectors)


cases = {'case 1' : {'u' :-1.25, 'v':-2.57, 'heading': -170},
         'case 2' : {'u' :-3.25, 'v':6.57,  'heading': -170},
         'case 3' : {'u' :1.25,  'v':3.57,  'heading': -170},
         'case 4' : {'u' :-1.25, 'v':-2.57, 'heading': -10},
         'case 5' : {'u' :-3.25, 'v':6.57,  'heading': -10},
         'case 6' : {'u' :1.25 , 'v':3.57,  'heading': -10},
        }
fig = plt.figure(dpi=120,figsize=(11,12))
gs = fig.add_gridspec(
        3,
        2,
        left=0.1,
        right=0.9,
        bottom=0.1,
        top=0.9,
        wspace=0.05,
        hspace=0.15,
    )
cpt_ax = 0

for case in cases:
    
    ground_heading_angle = cases[case]['heading']
    u_component = cases[case]['u']
    v_component = cases[case]['v']
    az,azscat,azscat180 = aar.compute_angles.azimuth_direction(ground_heading_angle,u_component,v_component)
    antenna_pointing_direction = ground_heading_angle+90 # north clock
    # print('antenna_pointing_direction',antenna_pointing_direction)
    wdir_wrt_north = aar.compute_angles.wind_direction_from_u_and_v(u_component,v_component)%360
    # print('wdir_wrt_north',wdir_wrt_north)
    # testmyarc = Arc((0,0), 2 * 1, 2 * 1, angle=-40,
    #                theta1=wdir_wrt_north, theta2=antenna_pointing_direction, color='green',
    #                label=r'${:.4}^\circ$'.format(float(az)))
    
    wind_to_line,heading_line,antenna = aar.draw_angle_between_vectors.create_line_from_uv(u_component,v_component,ground_heading_angle)
    
    # angle_plot_azimuth = aar.draw_angle_between_vectors.get_angle_plot(antenna,wind_to_line, 1)
    if az<0:
        angle_plot_azimuth = aar.draw_angle_between_vectors.get_angle_plot(antenna,wind_to_line, 1,keep_order=True,angle_expected=az)
    else:
        angle_plot_azimuth = aar.draw_angle_between_vectors.get_angle_plot(wind_to_line,antenna, 1,keep_order=True,angle_expected=az)
    angle_text_azimuth = aar.draw_angle_between_vectors.get_angle_text(angle_plot_azimuth)
    # print(angle_text_azimuth)
    # fig = plt.figure(dpi=120,figsize=(8,6))
    # ax = fig.add_subplot(1, 1, 1)
    # ax = gs[cpt_ax].gca()
    cpt_ax_i = cpt_ax%3
    cpt_ax_j = cpt_ax%2
    ax = fig.add_subplot(gs[cpt_ax_i,cpt_ax_j])
    
    
    # ax.add_line(north_line)
    ax.add_line(heading_line)
    ax.add_line(wind_to_line)
    ax.add_line(antenna)
    ax.add_patch(angle_plot_azimuth)  # To display the angle arc
    # ax.add_patch(testmyarc)
    ax.text(*angle_text_azimuth)  # To display the angle value
    plt.grid(True)
    
    ax.set_xlim(-3, 3)
    ax.set_ylim(-2, 2)
    plt.title('%s expected azimuth value : %1.2f°'%(case,az))
    unorth = np.array([0])
    vnorth = np.array([4.5])
    xnorth = np.array([0])
    ynorth = np.array([0])
    north_line = ax.quiver(xnorth,ynorth, unorth,vnorth,linewidth=3,scale=20, linestyle="-", color="black",label='North')
    # print(north_line)
    # ax.quiverkey(north_line, X=0.3, Y=1.1, U=10,
    #              label='North Quiver key, length = 10', labelpos='N')
    cpt_ax += 1
    plt.legend()
    plt.minorticks_on()
    # plt.show()
```
